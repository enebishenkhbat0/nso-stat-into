```python

import pandas as pd
from google.colab import files
from IPython.display import display

# 1. Excel файл оруулах
uploaded = files.upload()
filename = next(
    (f for f in uploaded if f.endswith(".xlsx")),
    None
)
if filename is None:
    raise ValueError("Excel (.xlsx) файл оруулна уу.")

df = pd.read_excel(
    filename,
    dtype={"Cadastre": "string"}
)

# 2. Өгөгдөл цэвэрлэх
df["Cadastre"] = (
    df["Cadastre"].astype("string")
    .str.strip()
    .str.replace(r"\.0$", "", regex=True)
)

df["NAME"] = df["NAME"].astype("string").str.strip()
df["Year"] = pd.to_numeric(df["Year"], errors="coerce")

area = "Тариалсан талбай_Total"
df[area] = pd.to_numeric(df[area], errors="coerce")

# 3. Зөвхөн 2024 оны тариалалт
data = df[
    (df["Year"] == 2024) &
    (df[area] > 0) &
    df["Cadastre"].notna() &
    df["Cadastre"].ne("") &
    df["NAME"].notna() &
    df["NAME"].ne("")
].copy()

# 4. CAD + таримал бүрийн давхардлыг нэгтгэх
# Давхар бичлэгийг давхар талбай гэж тооцохгүй.
crops = (
    data.groupby(["Cadastre", "NAME"], as_index=False)
    .agg(area_ha=(area, "max"))
)

# 5. Ерөнхий ангиллыг тэмдэглэх
general = {
    "Хүнсний ногоо", "Үр тариа",
    "Таримал тэжээл", "ногоон тэжээл",
    "Техникийн ургамал", "олон наст",
    "Бусад", "бусад"
}

crops["specific"] = ~crops["NAME"].isin(general)

# 6. CAD бүрт зонхилох таримал сонгох
crops = crops.sort_values(
    ["Cadastre", "specific", "area_ha", "NAME"],
    ascending=[True, False, False, True]
)

selected = crops.drop_duplicates(
    "Cadastre"
).copy()

# Нийт нарийвчилсан таримлын талбай
specific = crops[crops["specific"]]

totals = specific.groupby("Cadastre").agg(
    total_specific_area=("area_ha", "sum"),
    crop_count=("NAME", "nunique")
).reset_index()

selected = selected.merge(
    totals, on="Cadastre", how="left"
)

selected["dominance_pct"] = (
    selected["area_ha"] /
    selected["total_specific_area"] * 100
).round(2)

# 7. Шошгын найдвартай байдал
selected["label_status"] = "ШАЛГАХ"

selected.loc[
    selected["specific"] &
    (selected["crop_count"] == 1),
    "label_status"
] = "НЭГ ТАРИМАЛ"

selected.loc[
    selected["specific"] &
    (selected["dominance_pct"] >= 70) &
    (selected["crop_count"] > 1),
    "label_status"
] = "ЗОНХИЛОХ"

selected.loc[
    ~selected["specific"],
    "label_status"
] = "ЕРӨНХИЙ АНГИЛАЛ"

# 8. Байршлын мэдээлэл
locations = data.drop_duplicates("Cadastre")[
    ["Cadastre", "Dim_Aimag_Name", "Dim_Soum_Name"]
]

selected = selected.merge(
    locations, on="Cadastre", how="left"
)

selected = selected.rename(columns={
    "NAME": "crop_label",
    "area_ha": "main_crop_area"
})

# 9. RF сургалтад нэр дэвших 4 таримал
target_crops = ["Төмс", "Лууван", "Буудай", "Рапс"]

candidates = selected[
    selected["crop_label"].isin(target_crops) &
    selected["label_status"].isin([
        "НЭГ ТАРИМАЛ", "ЗОНХИЛОХ"
    ])
].copy()

# 10. Console-ийн үр дүн
print("===== 01. ЭХ ӨГӨГДӨЛ =====")
print("Нийт мөр:", len(df))
print("2024 оны зөв мөр:", len(data))
print("Ялгаатай CAD:", selected["Cadastre"].nunique())

print("\n===== 02. ШОШГЫН ЧАНАР =====")
print(selected["label_status"].value_counts())

print("\n===== 03. СУРГАЛТЫН 4 ТАРИМАЛ =====")
print(candidates["crop_label"].value_counts())

print("\n===== 04. СУРГАЛТЫН КАДАСТР =====")
print("Нэр дэвших CAD:", len(candidates))
print("Аймаг:", candidates["Dim_Aimag_Name"].nunique())
print(
    "Сум:",
    candidates[
        ["Dim_Aimag_Name", "Dim_Soum_Name"]
    ].drop_duplicates().shape[0]
)

display(candidates[[
    "Cadastre", "Dim_Aimag_Name",
    "Dim_Soum_Name", "crop_label",
    "dominance_pct", "label_status"
]].head(10))

# 11. Эх өгөгдлийг алдахгүй хадгалах
selected.to_csv(
    "CAD_2024_All.csv",
    index=False, encoding="utf-8-sig"
)

candidates.to_csv(
    "CAD_2024_Training_Candidates.csv",
    index=False, encoding="utf-8-sig"
)

print("\n===== 05. БЭЛЭН =====")
print("CAD_2024_All.csv")
print("CAD_2024_Training_Candidates.csv")
print("Анхаар: CAD ID-г GEE-тэй хараахан тулгаагүй.")
```
